# Local setup and F1 sign-in

Run once using **Python (primusEnv)** before the F1 year notebooks if needed. Weather notebooks need no F1 sign-in. This does not download race data, install packages or upgrade the environment. Credentials are entered through local input/hidden password prompts, never saved in notebook source/output. FastF1 manages its sign-in token in its own local user-data directory. Clear notebook outputs before sharing. If account access expires during a long batch, rerun this notebook and resume the year notebook.


```python
from pathlib import Path
import sys
import json
from IPython.display import display
import pandas as pd

ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents)
             if (p / 'data/archive/store.py').is_file() and (p / 'primusEnv').is_dir()), None)
if ROOT is None:
    raise RuntimeError('Open this notebook inside the F1 Primus AI project.')
expected = (ROOT / 'primusEnv/Scripts/python.exe').resolve()
if Path(sys.executable).resolve() != expected:
    raise RuntimeError(f'Select the primusEnv kernel: {expected}')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print('Python:', sys.executable)
```

```python
DRY_RUN = False
LOGIN = True
```

```python
from importlib.metadata import version
from packaging.version import Version
installed = Version(version('fastf1'))
if not Version('3.9') <= installed < Version('4'):
    raise RuntimeError('This collection expects FastF1 >=3.9,<4 in primusEnv. No automatic upgrade was performed.')
print('FastF1:', installed)
if LOGIN and not DRY_RUN:
    from data.archive.notebook_f1 import authenticate_in_notebook, collection_http, waiting_lock
    with waiting_lock(ROOT / '.local/f1_collection_control/authentication'):
        authenticate_in_notebook()
        collection_http(ROOT).reset_denied('fastf1')
    print('Local sign-in finished. Existing request history and cooldowns were retained.')
else:
    print('Sign-in disabled; no credentials requested.')
```
